```python
import os
import traci
```

```python
os.environ['SUMO_HOME'] = 'C:\Program Files (x86)\Eclipse\Sumo'

```

```python
sumo_binary = os.path.join(os.environ["SUMO_HOME"], "bin", "sumo-gui")  # Use 'sumo-gui' instead of 'sumo'
sumo_cmd = [sumo_binary, "-c", "test1.sumocfg"]
```

```python
# Start SUMO simulation
traci.start(sumo_cmd)

step = 0
while step < 1000:
    traci.simulationStep()
    
    vehicle_id = "t_2"  # Replace with an actual vehicle ID from your SUMO simulation
    
    if vehicle_id in traci.vehicle.getIDList():  # Check if the vehicle exists in the simulation
        speed = traci.vehicle.getSpeed(vehicle_id)  # Get current speed of the car
        
        # Simple speed control logic
        if speed < 5:  # If speed is too low, accelerate
            traci.vehicle.setAcceleration(vehicle_id, 2.0)  # Accelerate by 2 m/s²
        elif speed > 10:  # If speed is too high, decelerate
            traci.vehicle.setDecel(vehicle_id, 2.0)  # Decelerate by 2 m/s²

    step += 1

traci.close()
```
